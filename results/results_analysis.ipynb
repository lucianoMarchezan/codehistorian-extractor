```python
import json
import pandas as pd
import numpy as np

# Path to your JSON results file
json_file = "calc-test_python_detailed_results.json"

with open(json_file, "r", encoding="utf-8") as f:
    data = json.load(f)

df = pd.DataFrame(data)

print("=" * 60)
print("MODEL EVALUATION STATISTICS")
print("=" * 60)

print(f"Total records: {len(df):,}")

print("\nModels:")
print(df["model"].value_counts())

print("\nDatasets:")
print(df["dataset"].value_counts())

print("\nLanguages:")
print(df["language"].value_counts())

# Similarity stats
print("\n--- Embedding Similarity (sim) ---")
print(df["sim"].describe())

print("\n--- CodeBLEU ---")
print(df["codebleu"].describe())

# Correlation
print("\n--- Correlation ---")
corr = df[["sim", "codebleu"]].corr().iloc[0, 1]
print(f"Pearson correlation (sim vs codebleu): {corr:.4f}")

# Optional: check missing values
print("\nMissing values:")
print(df.isnull().sum())

# Top / bottom examples
print("\nTop 5 highest similarity pairs:")
display(df.sort_values("sim", ascending=False).head(5))

print("\nBottom 5 lowest similarity pairs:")
display(df.sort_values("sim", ascending=True).head(5))

print("\nTop 5 highest CodeBLEU pairs:")
display(df.sort_values("codebleu", ascending=False).head(5))
```
